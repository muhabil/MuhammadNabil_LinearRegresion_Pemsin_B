```python
# Muhammad Nabil 202431173

from sklearn.model_selection import train_test_split as tts
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Muhammad Nabil 202431173

df = pd.read_csv("botak.csv")
df
```

```python
# Muhammad Nabil 202431173

df.info()
```

```python
# Muhammad Nabil 202431173

import seaborn as sns

num_df = df.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(10,6))
sns.heatmap(num_df.corr(), annot=True, cmap="viridis")
plt.show
```

```python
# Muhammad Nabil 202431173

db = df.dropna(subset=["umur","gaji","tinggi","berat", "botak_prob"])

x = db[["umur","gaji","tinggi","berat"]]
y = db["botak_prob"]
```

```python
# Muhammad Nabil 202431173

x_train, x_test, y_train, y_test = tts(x,y,test_size=0.2, random_state=0)
```

```python
# Muhammad Nabil 202431173

model = LinearRegression()
model.fit(x_train, y_train)
```

```python
# Muhammad Nabil 202431173

y_pred = model.predict(x_test)
```

```python
# Muhammad Nabil 202431173

print("Coefficient : ", model.coef_)
print("Intercept :", model.intercept_)
print("MSE : ", mean_squared_error(y_test, y_pred))
print("R2 score :", r2_score(y_test, y_pred))
```

```python
# Muhammad Nabil 202431173

plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred, alpha = 0.5)
plt.plot([0,1],[0,1], 'r--')
plt.xlabel('Nilai asli y_test')
plt.ylabel('Nilai asli y_pred')
plt.title("Nilai asli vs nilai prediksi")
plt.show()
```
